# 02 — Tipus de dades

Python té diversos tipus de dades. En aquest notebook veurem nombres enters i decimals, nombres complexos, cadenes de text, valors booleans i `NoneType`.

## Nombres enters: `int`

Els enters són nombres sense part decimal. Python també permet treballar amb nombres enters molt grans.

````python
print("int:")  # Enters (números sense part decimal)
print(type(10))  # Nombre enter positiu
print(type(0))  # El número zero també és un enter
print(type(-5))  # Nombre enter negatiu
print(type(7238424723784278934789239874))  # Python permet nombres enters molt grans
````

## Nombres decimals: `float`

Els nombres `float` tenen part decimal. També es poden escriure amb notació científica.

````python
print("float:")  # Nombres decimals (de coma flotant)
print(type(3.14))  # Nombre amb part decimal
print(type(1.0))  # També és un float, encara que el valor sigui enter
print(type(1e3))  # Notació científica (equivalent a 1000.0)
````

## Nombres complexos: `complex`

Un nombre complex té una part real i una part imaginària. En Python, la part imaginària s'escriu amb `j`.

````python
print("complex:")  # Nombres complexos (amb part real i imaginària)
print(type(1 + 2j))  # 1 és la part real i 2j és la part imaginària
````

## Cadenes de text: `str`

Les cadenes de text poden contenir lletres, nombres o altres caràcters. Les cometes triples permeten escriure cadenes que ocupen diverses línies.

````python
print("str:")  # Cadenes de text (strings)
print(type("Hola"))  # Una cadena amb text
print(type(""))  # Una cadena buida
print(type("123"))  # Entre cometes, és una cadena i no un nombre
print(type(""" 
  Multilínia
"""))  # Una cadena de diverses línies
````

## Valors booleans: `bool`

Els valors booleans només poden ser `True` (cert) o `False` (fals). Les comparacions també retornen un valor booleà.

````python
print("bool:")  # Valors booleans (True o False)
print(type(True))  # Valor booleà cert
print(type(False))  # Valor booleà fals
print(type(1 < 2))  # Comparació que retorna un booleà (True)
````

## Absència de valor: `NoneType`

`None` representa l'absència d'un valor.

````python
print("NoneType:")  # Representa l'absència de valor
print(type(None))  # None representa "sense valor" o "nul"
````

# 03 — Conversió de tipus (*casting*)

La conversió de tipus és el procés de convertir un valor d'un tipus de dada a un altre. És útil quan cal fer operacions entre dades de tipus diferents. En aquests exemples convertirem nombres enters, decimals, cadenes de text i valors booleans.

## Conversió de cadena a enter

No es pot sumar directament una cadena i un enter; primer convertim la cadena amb `int()`.

````python
print(2 + int("100"))  # Converteix "100" a enter i hi suma 2. Resultat: 102
````

Aquesta suma provocaria un `TypeError`, perquè no es pot sumar directament un enter i una cadena:

```python
print("100" + 2)
```

## Conversió d'enter a cadena

La funció `str()` converteix un nombre en cadena de text, que es pot concatenar amb una altra cadena.

````python
print("100" + str(2))  # Converteix 2 a cadena i el concatena. Resultat: "1002"
````

## Conversió a decimal

La funció `float()` converteix una cadena amb un nombre decimal en un valor de tipus `float`.

````python
print(type(float("3.1416")))  # Converteix "3.1416" a float
````

## Conversió de decimal a enter

En convertir un `float` a `int`, s'elimina la part decimal.

````python
print(int(3.1416))  # Resultat: 3
````

## Conversió a booleà

En aquests exemples, el zero i la cadena buida es converteixen en `False`. Un nombre diferent de zero o una cadena no buida es converteixen en `True`.

````python
print(bool(3))  # True
print(bool(0))  # False
print(bool(-1))  # True
print(bool(""))  # False
print(bool(" "))  # True
print(bool("False"))  # True: la cadena no és buida
````

## Arrodoniment

La funció `round()` arrodoneix un nombre decimal.

````python
print(round(2.51))  # Resultat: 3
````

## Una conversió que provoca un error

A Python se sol fer servir `snake_case`: paraules en minúscula separades per guions baixos. Els noms en majúscules se solen reservar per a constants, tot i que Python no les fa constants de manera efectiva.

````python
nom, edat, ciutat = "pepito", 32, "Madrid"

el_meu_nom_de_variable = "correcte"  # snake_case
nom = "correcte"

elMeuNomDeVariable = "no recomanat"  # camelCase
ElMeuNomDeVariable = "no recomanat"  # PascalCase
elmeunomdevariable = "no recomanat"  # tot junt

el_meu_nom_de_variable_123 = "correcte"

LA_MEVA_CONSTANT = 3.14  # UPPER_CASE -> constants
LA_MEVA_CONSTANT = 3  # Python no té constants, però se segueix aquesta convenció
````

Un nom de variable no pot començar amb un nombre ni contenir guions o espais. Tampoc no es poden fer servir paraules reservades de Python com a noms de variables. Aquests exemples provocarien errors i, per tant, no s'executen:

```python

el meu nom = "incorrecte"  # No pot contenir espais
True = False  # No es poden sobreescriure les paraules reservades
```

Entre les paraules reservades de Python hi ha `False`, `None`, `True`, `and`, `as`, `assert`, `async`, `await`, `break`, `class`, `continue`, `def`, `del`, `elif`, `else`, `except`, `finally`, `for`, `from`, `global`, `if`, `import`, `in`, `is`, `lambda`, `nonlocal`, `not`, `or`, `pass`, `raise`, `return`, `try`, `while`, `with` i `yield`.

# 01 — Sentències condicionals (`if`, `elif`, `else`)

Les sentències condicionals permeten executar blocs de codi només si es compleixen certes condicions.

## `if`: executar codi si es compleix una condició

Si la condició és certa, s'executa el bloc indentat. Si no, aquest bloc s'omet.

````python
edat = 18
if edat >= 18:
    print("Ets major d'edat")
    print("Felicitats!")
````

## `else`: alternativa quan la condició no es compleix

El bloc `else` s'executa quan la condició de l'`if` és falsa.

````python
edat = 15
if edat >= 18:
    print("Ets major d'edat")
else:
    print("Ets menor d'edat")
````

## `elif`: comprovar diverses condicions

Es comproven les condicions en ordre i només s'executa el primer bloc que en compleix una. Si cap condició és certa, s'executa `else` (si n'hi ha).

````python
nota = 7
if nota >= 9:
    print("Excel·lent!")
elif nota >= 7:
    print("Notable!")
elif nota >= 5:
    print("Aprovat!")
else:
    print("No ha aprovat!")
````

## Condicions múltiples i operadors lògics

- `and`: cert si tots dos operands són certs.
- `or`: cert si almenys un operand és cert.
- `not`: nega una condició.

En JavaScript, `&&` equival a `and` i `||` equival a `or`.

````python
edat = 16
te_carnet = True

if edat >= 18 and te_carnet:
    print("Pots conduir 🚗")
else:
    print("POLICIA 🚔!!!1!!!")

# Amb or n'hi hauria prou de complir una de les dues condicions:
if edat >= 18 or te_carnet:
    print("Pots conduir a l'illa Margarita 🚗")

es_cap_de_setmana = False
if not es_cap_de_setmana:
    print("Avui hi ha classe")
````

## Condicionals imbricats

Es poden posar condicionals dins d'altres condicionals per comprovar diverses condicions. Sovint es pot expressar la mateixa lògica amb una estructura més plana i fàcil de llegir.

````python
edat = 20
te_diners = False

# Condicionals imbricats
if edat >= 18:
    if te_diners:
        print("Pots anar a la discoteca")
    else:
        print("Pots entrar a la discoteca, però no comprar res")
else:
    print("No pots entrar a la discoteca")

# Una alternativa amb menys imbricació
if edat < 18:
    print("No pots entrar a la discoteca")
elif te_diners:
    print("Pots anar a la discoteca")
else:
    print("Queda't a casa")
````

## Valors interpretats com a certs o falsos

A més de `True` i `False`, alguns valors es poden utilitzar directament com a condicions. Per exemple, un nombre diferent de zero es considera cert, mentre que `0` i la cadena buida `""` es consideren falsos.

````python
numero = 5
if numero:
    print("El nombre no és zero")

numero = 0
if numero:
    print("Aquest missatge no es mostrarà")
print("Aquest missatge sí que s'executa")

nom = ""
if nom:
    print("El nom no és buit")
````

## Assignació i comparació

No confonguis `=` (assignació d'un valor) amb `==` (comparació d'igualtat).

````python
numero = 3  # Assignació
es_el_tres = numero == 3  # Comparació: True

if es_el_tres:
    print("El nombre és 3")
````

## Expressió condicional

Una expressió condicional permet triar entre dos valors en una sola línia. En JavaScript hi ha una forma equivalent amb l'operador ternari `? :`.

````python
edat = 197
missatge = "És major d'edat" if edat >= 18 else "És menor d'edat"
print(missatge)
````

# 02 — Valors booleans

# 03 — Llistes

Les llistes són seqüències mutables d'elements i poden contenir dades de tipus diferents.

## Crear llistes

Una llista s'escriu entre claudàtors. Pot estar buida, contenir elements del mateix tipus o combinar tipus diferents. També es poden crear llistes imbricades i matrius.

````python
llista1 = [1, 2, 3, 4, 5]  # Llista d'enters
llista2 = ["pomes", "peres", "plàtans"]  # Llista de cadenes
llista3 = [1, "hola", 3.14, True]  # Llista de tipus diferents
llista_buida = []
llista_de_llistes = [[1, 2], ["mitjó", 4]]
matriu = [[1, 2], [2, 3], [4, 5]]

print(llista1)
print(llista2)
print(llista3)
print(llista_buida)
print(llista_de_llistes)
print(matriu)
````

## Accedir als elements mitjançant l'índex

Els índexs comencen per `0`. Els índexs negatius compten des del final: `-1` és l'últim element. En una llista imbricada, cal indicar un índex per a cada nivell.

````python
fruites = ["pomes", "peres", "plàtans", "maduixes", "kiwi"]
print(fruites[0])  # pomes
print(fruites[1])  # peres
print(fruites[2])  # plàtans
print(fruites[3])  # maduixes
print(fruites[-1])  # kiwi: últim element
print(fruites[-2])  # maduixes

llista_de_llistes = [[1, 2], ["mitjó", 4]]
print(llista_de_llistes[1][0])  # mitjó
print(llista_de_llistes[0][1])  # 2
print(llista_de_llistes[1][1])  # 4
````

## Seleccionar fragments (*slicing*)

La sintaxi `llista[inici:final]` selecciona elements des de l'índex inicial fins abans de l'índex final. Si s'omet un límit, s'utilitza el principi o el final de la llista. `[:]` en crea una còpia superficial.

````python
nombres_tallats = [1, 2, 3, 4, 5]
print(nombres_tallats[1:4])  # [2, 3, 4]
print(nombres_tallats[:3])  # [1, 2, 3]
print(nombres_tallats[3:])  # [4, 5]
print(nombres_tallats[:])  # [1, 2, 3, 4, 5]
nombres_copiats = nombres_tallats[:]  # Copia completa de la lista
primers_3_nombres=nombres_tallats[:3]  # [1, 2, 3]
els_ultims_3_nombres=nombres_tallats[-3:]  # [3, 4, 5]
````

El *slicing* també accepta un pas: `llista[inici:final:pas]`. Un pas negatiu permet recórrer la llista en ordre invers.

````python
llista1 = [1, 2, 3, 4, 5, 6, 7, 8]
print(llista1[1:6:2])  # [2, 4, 6]
print(llista1[::2])  # Elements dels índexs parells (posicions senars)
print(llista1[::-1])  # Elements en ordre invers
````

## Modificar una llista

Les llistes són mutables: es pot canviar el valor d'un element assignant-ne un de nou a un índex existent. No es poden afegir elements utilitzant un índex que encara no existeix: això provoca un `IndexError`.

````python
llista1 = [1, 2, 3, 4, 5, 6, 7, 8]
llista1[0] = 20
print(llista1)
````

Aquest accés no s'executa perquè l'índex `10` no existeix en aquesta llista i provocaria un `IndexError`:

```python
llista1[10] = 100
```

## Afegir elements i consultar la llargada

Es poden combinar llistes amb `+` o ampliar una llista amb `+=`. La funció `len()` retorna el nombre d'elements. En JavaScript, la llargada d'un array s'obté amb `.length`.

````python
llista1 = [1, 2, 3]

# Forma llarga
llista1 = llista1 + [4, 5, 6]
print(llista1)

# Forma curta
llista1 += [7, 8, 9]
print(llista1)

print("Llargada de la llista:", len(llista1))
````

# 04 — Mètodes de les llistes

En aquest notebook veurem mètodes per afegir, eliminar, ordenar i consultar els elements d'una llista.

## Afegir i inserir elements

- `append()` afegeix un element al final.
- `insert()` insereix un element a la posició indicada.
- `extend()` afegeix al final tots els elements d'un iterable.

````python
llista1 = ['a', 'b', 'c', 'd']
print(llista1)

# Afegir o inserir elements a la llista
# llista1[4] = 'e'
llista1.append('e')
print(llista1)

llista1.insert(1, '@')
print(llista1)

llista1.extend(['😃', '😍'])
# llista1 = llista1 + ['😃', '😍']
print(llista1)
````

## Eliminar elements

- `remove(valor)` elimina la primera aparició del valor.
- `pop()` elimina i retorna l'últim element; `pop(índex)` elimina i retorna l'element d'aquell índex.
- `del` elimina un element o un fragment.
- `clear()` buida la llista.

````python
llista1.remove('@')
print(llista1)

ultim = llista1.pop()
print(ultim)
print(llista1)

lletra_b = llista1.pop(1)
print(lletra_b)
print(llista1)

del llista1[-1]
print(llista1)

llista1.clear()
print(llista1)
````

Per eliminar un interval, es pot combinar `del` amb *slicing*.

````python
llista1 = ['🐼', '🐨', '🐶', '😿', '🐹']
del llista1[1:3]
print(llista1)
````

## Ordenar llistes

`sort()` ordena la llista original. `sorted()` retorna una llista ordenada nova i deixa intacta l'original.

````python
print("Ordenar la llista original")
nombres = [3, 10, 2, 8, 99, 101]
nombres.sort()
print(nombres)

print("\nCrear una llista ordenada nova")
nombres = [3, 10, 2, 8, 99, 101]
nombres_ordenats = sorted(nombres)
print(nombres)  # La llista original no s'ha modificat
print(nombres_ordenats)
````

Les cadenes s'ordenen alfabèticament. Amb `key=str.lower`, l'ordenació no distingeix entre majúscules i minúscules.

````python
print("Ordenar cadenes en minúscula")
fruites = ['Poma', 'pera', 'Llimona', 'poma', 'pera', 'llimona']
fruites_ordenades = sorted(fruites)
print(fruites_ordenades)

print("\nOrdenar cadenes sense distingir majúscules i minúscules")
fruites = ['poma', 'Pera', 'Llimona', 'Poma', 'pera', 'llimona']
fruites.sort(key=str.lower)
print(fruites)
````

## Consultar llistes

`len()` retorna la llargada, `count()` compta les aparicions d'un valor i l'operador `in` comprova si un element és present.

````python
animals = ['🐶', '🐼', '🐨', '🐶']
print(len(animals))  # Mida de la llista -> 4
print(animals.count('🐶'))  # Aparicions de 🐶 -> 2
print('🐼' in animals)  # True
print('🐹' in animals)  # False
````

# 01 · Bucles `while`

Permeten **executar un bloc de codi repetidament mentre es compleixi una condició**.

> 💡 Recorda: la **indentació** (sagnat) defineix quin codi pertany al bucle.

## 🔁 Bucle `while` bàsic

És molt important **modificar el comptador** dins del bucle per evitar un bucle infinit.

````python
comptador = 0

while comptador <= 5:
    print(comptador)
    comptador += 1  # evita un bucle infinit
````

## 🛑 `break`: trencar el bucle

`break` surt del bucle immediatament.

````python
comptador = 0

while True:
    print(comptador)
    comptador += 1
    if comptador == 5:
        break  # surt del bucle
# print("Hola") Fora del while
````

## ⏭️ `continue`: saltar una iteració

`continue` salta la iteració actual i continua amb la següent.

````python
comptador = 0
while comptador < 10:
    comptador += 1

    if comptador % 2 == 0:
        continue

    print(comptador)
````

## ✅ `else` en un `while`

El bloc `else` s'executa quan el bucle **acaba de forma natural** (sense `break`).

````python
comptador = 0
while comptador < 5:
    print(f"El comptador és {comptador}")
    comptador += 1
else:
    print("El bucle while ha acabat de forma natural.")
    print(f"El valor final del comptador és {comptador}.")
````

Si el bucle s'interromp amb `break`, **l'`else` no s'executa**.

````python
numero = 0
while numero < 10:
    print(f"El número és {numero}")
    if numero == 3:
        print("Hem arribat al 3! Trencant el bucle.")
        break  # interromp el bucle
    numero += 1
else:
    print("Aquest missatge no s'imprimirà perquè el bucle ha estat interromput per 'break'.")

print("El programa ha continuat després del bucle.")
````

## ⌨️ Validar l'entrada de l'usuari

Demanem un número fins que sigui positiu.

````python
numero = -1
while numero < 0:
    numero = int(input("Escriu un número positiu: "))
    if numero < 0:
        print("El número ha de ser positiu. Torna-ho a provar.")

print(f"El número que has introduït és {numero}")
````

Amb `try` / `except` controlem que l'usuari realment escrigui un número.

````python
numero = -1
while numero < 0:
    try:
        numero = int(input("Escriu un número positiu: "))
        if numero < 0:
            print("El número ha de ser positiu. Torna-ho a provar.")
    except:
        print("El que introdueixes ha de ser un número, que si no peta!")

print(f"El número que has introduït és {numero}")
````
